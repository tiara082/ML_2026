## Part 2 - Ethics, Law, Energy, and the Environment

### A. Examples of unethical or illegal uses of AI

1. **Creating deepfakes for fraud or defamation.** Changing someone's face or voice without permission can mislead people and damage that person's reputation. It can also violate privacy and electronic information laws.
2. **Using personal data without permission.** Collecting face photos, voice recordings, location data, or health information to train a model without a clear purpose or proper protection is a serious privacy problem.
3. **Making discriminatory automated decisions.** A hiring, credit, or risk assessment model can treat some groups unfairly if it learns from biased data. Important decisions should be checked by people instead of being accepted blindly.
4. **Copying someone's work or identity without permission.** Generating text, images, or voices that imitate another person's work or identity without credit or permission can raise copyright and personality-rights issues.
5. **Using AI for cyberattacks and disinformation.** AI can be used to create phishing messages, malware, fake news, or content intended to manipulate elections. These uses can harm public safety and trust in information.

Some basic principles should be followed when using AI: be transparent, ask for consent, collect only the data that is needed, protect the data, check for bias, and make someone responsible for the result. For high-risk systems, the model should be tested and documented, and people should have a way to question or correct a decision.

### B. Energy and environmental impacts, and possible solutions

1. **Electricity used by data centers.** Training and running large models require GPUs or TPUs, as well as cooling systems. More frequent use means more electricity consumption and, in some places, more carbon emissions.
2. **Carbon emissions and water use.** When the electricity comes from fossil fuels, the carbon footprint is higher. Data centers may also use water for cooling, which can be a problem in areas that already have limited water supplies.
3. **Electronic waste.** The growth of AI increases the demand for chips, servers, and network equipment. Replacing this hardware too quickly can create more electronic waste and increase the need for mined materials.
4. **Building new infrastructure.** Data centers need land, construction materials, electricity, and network connections. Their environmental impact therefore comes from more than just the model training process.

**Possible ways to reduce the impact:**

- use a smaller and more efficient model when it is good enough;
- use transfer learning, caching, quantization, pruning, and batch inference;
- measure the energy and emissions of experiments, not just their accuracy;
- run workloads in places or at times when cleaner electricity is available;
- keep servers in use for longer and recycle old hardware properly;
- choose data centers that report their energy and water use;
- remove unnecessary experiments and files so that they do not continue using storage and computing resources.

**My opinion:** the biggest model is not always the best choice. For a class project or a simple application, a smaller model that gives acceptable results is often the more responsible option. It can save time and money while also reducing the environmental impact.
