In this lab, I fill in missing values in the Titanic dataset. The columns that need attention are `Age`, `Cabin`, and `Embarked`.

## Step 0 - Preparation

I use pandas to load the dataset and NumPy for numerical operations.

```python
%pip install -q pandas numpy

import pandas as pd
import numpy as np
```

## Step 1 - Loading the Data

The Titanic CSV file is stored in the same folder as this notebook.

```python
df = pd.read_csv("Titanic-Dataset.csv")

# Display the first five rows
df.head()
```

## Step 2 - Data Inspection

Before filling the missing values, I check the data types and the number of empty values.

```python
df.info()
```

```python
missing_before = df.isnull().sum()
missing_before[missing_before > 0]
```

## Step 3 - Data Imputation

I use a different method for each column:

- `Age` is numerical, so I use its mean value.
- `Cabin` has many missing values. I replace them with `DECK` to show that the cabin information is not available.
- `Embarked` has only a few missing values, so I use its most common value, or mode.

```python
# Age - replace missing values with the mean
age_mean = np.mean(df["Age"].dropna())
df["Age"] = df["Age"].fillna(age_mean)

# Cabin - use a label for passengers without cabin information
df["Cabin"] = df["Cabin"].fillna("DECK")

# Embarked - replace missing values with the mode
embarked_mode = df["Embarked"].mode()[0]
df["Embarked"] = df["Embarked"].fillna(embarked_mode)

df.head()
```

## Step 4 - Result Validation

I check the data again to make sure that the selected columns no longer contain missing values.

```python
df.info()
```

```python
missing_after = df[["Age", "Cabin", "Embarked"]].isnull().sum()
missing_after
```

```python
print("Mean age used:", age_mean)
print("Embarked value used for missing rows:", embarked_mode)
print("Number of DECK values:", (df["Cabin"] == "DECK").sum())
```

## Optional - Save the Result

The completed dataset can be saved as a new CSV file. The original Titanic file is not overwritten.

```python
df.to_csv("Titanic-Dataset-Fixed.csv", index=False)
print("Saved as Titanic-Dataset-Fixed.csv")
```
