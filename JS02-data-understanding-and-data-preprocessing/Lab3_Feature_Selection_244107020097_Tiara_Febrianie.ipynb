In this lab, I select useful features from the Titanic dataset, convert categorical values into numbers, and standardize the `Age` column.

## Overview

The features used in this example are `Survived`, `Pclass`, `Sex`, `Age`, and `Cabin`. The dataset was cleaned through the imputation process in Lab 2.

## Step 0 - Load the Libraries

I use pandas for the dataset. `LabelEncoder` changes text categories into numbers, while `StandardScaler` puts numerical values on a similar scale.

```python
%pip install -q pandas scikit-learn

import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Step 1 - Load the Data

The imputed Titanic dataset is stored in the same folder as this notebook.

```python
dpath = "Titanic-Dataset-Fixed.csv"
df = pd.read_csv(dpath)

df.head()
```

## Step 2 - Slice the Data

I keep only the variables used in this exercise. `Survived` is the target variable, and the other columns are the selected features.

```python
selected_columns = ["Survived", "Pclass", "Age", "Sex", "Cabin"]
df = df[selected_columns].copy()

df.head()
```

## Step 3 - Encoding

`Sex` and `Cabin` contain text values. I use label encoding to convert each category into an integer.

```python
sex_encoder = LabelEncoder()
cabin_encoder = LabelEncoder()

df["Sex"] = sex_encoder.fit_transform(df["Sex"])
df["Cabin"] = cabin_encoder.fit_transform(df["Cabin"])

df.head()
```

## Step 4 - Verify the Encoding Results

The categorical columns are now represented by integer values. The category mapping can be checked below.

```python
print("Sex mapping:")
print(dict(zip(sex_encoder.classes_, sex_encoder.transform(sex_encoder.classes_))))

print("Number of cabin categories:", len(cabin_encoder.classes_))
df.info()
```

## Step 5 - Standardization

I standardize `Age` with `StandardScaler`. This changes the values so that the column has a mean close to 0 and a standard deviation close to 1.

```python
age_scaler = StandardScaler()
df["Age"] = age_scaler.fit_transform(df[["Age"]])

df.head()
```

## Step 6 - Verify the Standardization Results

I check the mean and standard deviation to confirm the result.

```python
print("Mean of standardized Age:", df["Age"].mean())
print("Standard deviation of standardized Age:", df["Age"].std())

df.head()
```

## Conclusion

The selected variables are ready for machine learning. `Sex` and `Cabin` were encoded as numbers, and `Age` was standardized.
