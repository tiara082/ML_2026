In this lab, I perform simple preprocessing on an image. The steps are loading the image, resizing it to 128 x 128 pixels, and converting it to grayscale.

## Introduction

Image preprocessing is usually done before feature extraction or model training. Resizing makes images have the same dimensions, while grayscale reduces the image to one intensity channel.

## Step 0 - Preparation

I use OpenCV to read and process the image, and Matplotlib to display the results.

```python
%pip install -q opencv-python matplotlib

import cv2
import matplotlib.pyplot as plt
```

## Step 1 - Load the Image

The image file is named `Lenna.png` and is stored in the same folder as this notebook. OpenCV reads images in BGR order by default, so I convert the image to RGB before displaying it with Matplotlib.

```python
image_path = "Lenna.png"

# OpenCV reads images in BGR format by default
img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError(f"Could not find {image_path}")

# Convert BGR to RGB for correct display in Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

## Step 2 - Resize the Image

The image is resized to 128 x 128 pixels. Resizing can reduce the level of detail, but it makes the image size consistent for later processing.

```python
img_resized = cv2.resize(img_rgb, (128, 128))

plt.figure(figsize=(6, 6))
plt.imshow(img_resized)
plt.title("Image after Resizing (128 x 128)")
plt.axis("off")
plt.show()

print("New image shape:", img_resized.shape)
```

## Step 3 - Convert to Grayscale

The resized image is converted from RGB to grayscale. A grayscale image has one intensity value for each pixel instead of three color channels.

```python
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(6, 6))
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

print("Grayscale image shape:", img_gray.shape)
```

## Conclusion

The image was loaded, converted from BGR to RGB, resized to 128 x 128 pixels, and converted to grayscale. These are common first steps when preparing image data for feature extraction or machine learning.
