## Step 0 - Load the Libraries

I use pandas to work with the dataset and scikit-learn for label encoding and standardization.

```python
%pip install -q pandas scikit-learn

import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Step 1 - Load the Dataset

The dataset file is named `wbc.csv` and is stored in the same folder as this notebook.

```python
df = pd.read_csv("wbc.csv")

print("Dataset shape:", df.shape)
df.head()
```

## Step 2 - Identify Usable and Unusable Variables

The `id` column is only an identifier for each record. It does not describe the tumour, so it should not be used as a model feature.

The `diagnosis` column contains the target label. It is useful as the value the model tries to predict, but it is not used as an input feature.

The remaining 30 columns describe measurements such as radius, texture, perimeter, area, smoothness, and symmetry. These numerical columns can be used as features.

```python
target_column = "diagnosis"
id_column = "id"
feature_columns = [column for column in df.columns if column not in [id_column, target_column]]

print("Identifier column:", id_column)
print("Target column:", target_column)
print("Number of feature columns:", len(feature_columns))
print("Feature columns:")
print(feature_columns)
```

```python
df["diagnosis"].value_counts()
```

## Step 3 - Encode the Diagnosis Column

The diagnosis values are text labels: `M` means Malignant and `B` means Benign. I convert them into numbers so they can be used by a machine learning algorithm.

```python
diagnosis_encoder = LabelEncoder()
df["diagnosis"] = diagnosis_encoder.fit_transform(df["diagnosis"])

diagnosis_mapping = dict(zip(diagnosis_encoder.classes_, diagnosis_encoder.transform(diagnosis_encoder.classes_)))
print("Diagnosis mapping:", diagnosis_mapping)
df[["diagnosis"]].head()
```

## Step 4 - Verify the Encoding

The diagnosis column is now numeric. The original mapping is displayed so the encoded values can be interpreted correctly.

```python
print(df["diagnosis"].value_counts().sort_index())
print("Encoded data types:")
print(df[["id", "diagnosis"]].dtypes)
```

## Step 5 - Standardize the Numerical Features

I standardize the 30 numerical feature columns with `StandardScaler`. The `id` column is excluded because it is only an identifier. The encoded diagnosis is also excluded because it is the target label, not a measurement feature.

```python
feature_scaler = StandardScaler()
df[feature_columns] = feature_scaler.fit_transform(df[feature_columns])

df.head()
```

## Step 6 - Verify the Standardization

After standardization, each feature should have a mean close to 0 and a standard deviation close to 1.

```python
standardization_check = pd.DataFrame({
    "mean": df[feature_columns].mean(),
    "standard_deviation": df[feature_columns].std()
})

standardization_check.head()
```

## Optional - Save the Preprocessed Dataset

The processed data can be saved for use in later machine learning exercises.

```python
df.to_csv("wbc-preprocessed.csv", index=False)
print("Saved as wbc-preprocessed.csv")
```

## Conclusion

The `id` column was excluded because it is only an identifier. The `diagnosis` column was kept as the target and encoded into numbers. The 30 numerical measurement columns were standardized so they can be used more fairly by a machine learning model.
