This lab extracts text features from a small corpus using Term Frequency-Inverse Document Frequency (TF-IDF).

## Overview

Each sentence is treated as one document. The corpus is transformed into a numerical feature matrix, where each column represents a term and each value is that term's TF-IDF weight in a document.

## Step 0 - Prepare the Corpus

The corpus contains five short documents from the lab instructions.

```python
corpus = [
    'the house had a tiny little mouse',
    'the cat saw the mouse',
    'the mouse ran away from the house',
    'the cat finally ate the mouse',
    'the end of the mouse story'
]

print('Number of documents:', len(corpus))
for index, document in enumerate(corpus, start=1):
    print(f'Document {index}: {document}')
```

## Step 1 - Construct the TF-IDF Model

`TfidfVectorizer` removes English stop words and calculates the TF-IDF representation for every document.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vect = TfidfVectorizer(stop_words='english')
resp = vect.fit_transform(corpus)

print('Sparse TF-IDF matrix:')
print(resp)
print('Matrix shape:', resp.shape)
```

The sparse matrix output uses `(row, column)` coordinates. The row identifies a document, the column identifies a term, and the value is the TF-IDF weight for that term in the document.

## Step 2 - Inspect the Terms Used

The feature names show the vocabulary and the column order used by the TF-IDF matrix.

```python
feature_names = vect.get_feature_names_out()

print('Feature names:')
print(feature_names)
print('Number of features:', len(feature_names))
```

## Step 3 - View the Dense Feature Matrix


```python
import pandas as pd

tfidf_table = pd.DataFrame(
    resp.toarray(),
    columns=feature_names,
    index=[f'Document {index}' for index in range(1, len(corpus) + 1)]
)

tfidf_table
```

## Conclusion

The five text documents were converted into a 5 x 12 numerical feature matrix. Common words that are removed as English stop words do not become features, while the remaining terms receive TF-IDF weights that can be used by machine-learning models.
